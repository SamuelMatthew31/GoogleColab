```python
from PIL import Image, ImageFont, ImageDraw
from PIL.ImageChops import add, subtract
from skimage.io import imread, imshow, show
from skimage.transform import SimilarityTransform, warp, swirl
from scipy.ndimage import affine_transform
import matplotlib.pylab as plt
import matplotlib.image as mpimg
```

# MENGIMPOR DAN MENDEFINISIKAN GAMBAR

```python
import numpy as np
im1 = Image.open("/content/arduino.jpg")
im2 = Image.open("/content/rangkaian.jpg")

#Testing untuk melihat ratio resolusi pada gambar agar dapat memperkirakan perhitungannya
print(im1.width, im1.height, im1.mode, im1.format)
```

# BAGIAN A

```python
#Membuat gambar menjadi terputar balik
im1_rotate1 = im1.transpose(Image.FLIP_LEFT_RIGHT)
#Membuat gambar dapat dimiringkan sesuai dengan posisi yang diinginkan
im1_rotate30 = im1_rotate1.rotate(16)
#Membuat ukuran gambar menjadi lebih kecil
im1_size = im1_rotate30.resize((600,400))
#Membuat crop pada gambar sesuai dengan posisi gambar yang diinginkan
im1_crop = im1_size.crop((55,53,546,330))

#Mengetest ukuran pada gambar sekarang, setelah diproses
print(im1_size.width, im1_size.height, im1_size.mode, im1_size.format)

draw = ImageDraw.Draw(im1_crop)
draw.rectangle((30, 200, 180, 250), fill=(255, 255, 255, 255))  # Putih solid
draw.rectangle((35, 205, 175, 245), fill=(255, 255, 0, 255))  # Kuning solid
font = ImageFont.truetype("LiberationMono-Regular.ttf", 11) #use a truetype font
draw.text((40, 220), "Samuel Matthew-82363", font=font, fill=(0,0,0))
del draw

display(im1_crop)
```

# BAGIAN B

```python
#Mengubah gambar menjadi hitam putih menggunakan syntax convert()
im_grayscale = im1_crop.convert('L')

draw = ImageDraw.Draw(im1_crop)
draw.rectangle((30, 200, 180, 250), fill=(255, 255, 255, 255))  # Putih solid
draw.rectangle((35, 205, 175, 245), fill=(255, 255, 0, 255))  # Kuning solid
font = ImageFont.truetype("LiberationMono-Regular.ttf", 11) #use a truetype font
draw.text((40, 220), "Samuel Matthew-82363", font=font, fill=(0,0,0))
del draw

display(im_grayscale)
```

# BAGIAN C

```python
display(add(im1, im2))
fig, ax = plt.subplots(1, 3)
ax[0].imshow(im1)
ax[1].imshow(im1_crop)
ax[2].imshow(im_grayscale)
plt.show()
```
