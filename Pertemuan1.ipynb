```python
# 1. Import library utama
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 2. Upload gambar (opsi: dari local ke Colab)
from google.colab import files
uploaded = files.upload()

# Misal kita pakai nama file: "image.jpg"
img = cv2.imread("image.jpg")   # OpenCV baca dalam format BGR
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Konversi ke RGB (biar cocok dengan matplotlib)

# 3. Tampilkan gambar asli
plt.figure(figsize=(6,6))
plt.imshow(img_rgb)
plt.title("Gambar Asli")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(6,6))
plt.imshow(gray, cmap='gray')
plt.title("Grayscale")
plt.axis("off")
plt.show()
```

```python
resized = cv2.resize(img_rgb, (200, 200))
rotated = cv2.rotate(img_rgb, cv2.ROTATE_90_CLOCKWISE)

plt.subplot(1,2,1)
plt.imshow(resized)
plt.title("Resize 200x200")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(rotated)
plt.title("Rotate 90°")
plt.axis("off")
plt.show()
```

```python
blurred = cv2.GaussianBlur(img_rgb, (11, 11), 0)

plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("Asli")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(blurred)
plt.title("Gaussian Blur")
plt.axis("off")
plt.show()
```

```python
edges = cv2.Canny(gray, 100, 200)

plt.figure(figsize=(6,6))
plt.imshow(edges, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")
plt.show()
```

```python
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(6,6))
plt.imshow(thresh, cmap='gray')
plt.title("Thresholding")
plt.axis("off")
plt.show()
```
