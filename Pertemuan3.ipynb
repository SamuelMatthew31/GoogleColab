# CALCULATE AND PLOT HISTOGRAM OF AN IMAGE

```python
import matplotlib.pylab as plt
import numpy as np
import cv2
```

```python
im1 = cv2.imread('bunga.jpg',  0)
hist = cv2.calcHist([im1], [0], None, [256], [0, 256])
hist,bins = np.histogram(im1.ravel(),256,[0,256])
plt.plot(hist)
```

```python
im1 = cv2.imread('bunga.jpg')
color = ('b','g','r')
for i,col in enumerate(color):
    histr = cv2.calcHist([im1],[i],None,[256],[0,256])
    plt.plot(histr,color = col)
    plt.xlim([0,256])
plt.show()
```

# IMAGE SMOOTHING/IMAGE BLURRING

```python
im3 = cv2.imread('/content/merak.jpg')

kernel = np.ones((55,55),np.float32)/3025
dst = cv2.filter2D(im3,-1,kernel)

plt.subplot(121),plt.imshow(im3),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(dst),plt.title('Averaging')
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
im3 = cv2.imread('/content/merak.jpg')

blur = cv2.blur(im3,(55,55))

plt.subplot(121),plt.imshow(im3),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(blur),plt.title('Blurred')
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
im3 = cv2.imread('/content/merak.jpg')

blur = cv2.GaussianBlur(im3,(55,55),0)

plt.subplot(121),plt.imshow(im3),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(blur),plt.title('Blurred')
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
im3 = cv2.imread('/content/merak.jpg')

median = cv2.medianBlur(im3,55)

plt.subplot(121),plt.imshow(im3),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(blur),plt.title('Blurred')
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
im3 = cv2.imread('/content/merak.jpg')

bilateral = cv2.bilateralFilter(im3,55,155,155)

plt.subplot(121),plt.imshow(im3),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(bilateral),plt.title('Blurred')
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
from scipy import misc, ndimage
from matplotlib.pyplot import imread

im4 = imread('/content/merak.jpg')
k = 49 #7x7 kernel
im_box = ndimage.uniform_filter(im4, size=(k,k,1))
s = 7
t = (((k-1)/2)-0.5)/s  #truncate parameter value for a kxk gaussian kernel with sigma s
im_gauss = ndimage.gaussian_filter(im4, sigma=(s,s,0), truncate=t)
fig = plt.figure(figsize=(30,10))
plt.subplot(131),plt.imshow(im4),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(132),plt.imshow(im_box),plt.title('Box Filter')
plt.xticks([]), plt.yticks([])
plt.subplot(133),plt.imshow(im_gauss),plt.title('Gaussian Filter')
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
from skimage.filters import laplace
from skimage.color import rgb2gray

im5 = rgb2gray(imread('/content/merak.jpg'))
im6 = np.clip(laplace(im5) + im5, 0, 1)
plt.figure(figsize=(20,30))
plt.subplot(121),plt.imshow(im5, cmap="gray"),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(im6, cmap="gray"),plt.title('Sharpened Image')
plt.xticks([]), plt.yticks([])
plt.show()
```
